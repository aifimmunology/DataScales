```python
import hisepy as hp
```

### 1. BRI Dataset

```python
df = hp.list_files_in_project_store('cohorts')
```

```python
pattern1="tar"
pattern2="polonium-tin-curium"
filtered_df = df[df['name'].str.contains(pattern1) & df['name'].str.contains(pattern2)]
```

```python
hise_res = hp.reader.cache_files(filtered_df['id'].tolist())
```

```python
import tarfile
from pathlib import Path

bri_extract_dir = Path("/home/workspace/bri_h5ads")
bri_extract_dir.mkdir(exist_ok=True)

for tar_path in hise_res:
  with tarfile.open(tar_path) as tar:
      tar.extractall(bri_extract_dir)

bri_h5ads = sorted(bri_extract_dir.glob("sample_h5ad/*.h5ad"))
print(f"total BRI h5ads: {len(bri_h5ads)}")
```
